```python
import seaborn as sns
import pandas as pd
import cufflinks as cf
from plotly.offline import iplot
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
import plotly.express as px
import numpy as np

# Use px.line instead of the deprecated iplot method
fig = px.line(tips, y='total_bill')
fig
```

```python
# Prepare your data first
daily_avg = tips.groupby('day', observed=True)['tip'].mean().reset_index()

# Create the plot
fig = px.bar(daily_avg, x='day', y='tip', 
             title='Average Tip per Day',
             color_discrete_sequence=['#ff9933']) # This matches your orange color
fig.show()
```

```python
fig = px.scatter(tips, x='total_bill', y='tip', 
                 title='Total Bill vs Tip')

fig.show()
```

```python
fig = px.box(tips, x='day', y='total_bill', 
             title='Total Bill Distribution by Day',
             color='day')

fig.show()
```
