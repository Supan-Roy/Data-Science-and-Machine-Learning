```python
import numpy as np
```

ARRAY OPERATIONS

Arithmetic Operations

```python
a1 = np.array([1,2,3,4,5])
a2 = np.array([6,7,8,9,10])
```

```python
a1+a2
```

Broadcasting

```python
l = [10,20,30,40]
arr = np.array(l)
```

```python
arr+10
```

```python
arr2 = np.arange(1,26).reshape(5,5)
arr2
```

```python
arr2+10
```

Deep and Shallow Copy

```python
a = np.arange(1,21)
a
```

```python
slice = a[:5]
slice*5
```

```python
b = a.copy()
b
```

```python
b[0]=51
b
```

Matrix Operations

```python
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])
```

```python
A,B
```

```python
A@B
```

```python
np.dot(A,B)
```

```python
A.T
```

Stacking Array

```python
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
```

```python
np.vstack((a,b))
```

```python
np.hstack((a,b))
```

```python
np.column_stack((a,b))
```

Splitting Array

```python
c = np.arange(16).reshape(4,4)
c
```

```python
np.hsplit(c,2)
```

```python
d = np.vsplit(c,2)
```

```python
for i in d:
    print(i)
```
